# Example 1 — SBA State Atlas

Load the official SBA State Small Business Statistics workbook and prepare a state-level measure for mapping.


```python
from capital_access_atlas import (
    detect_state_column,
    load_sba_state_workbook,
    numeric_metric_columns,
    prepare_state_metric,
)
```

```python
metadata, sheets = load_sba_state_workbook()
metadata
```

```python
state_sheets = {}
for name, frame in sheets.items():
    state_column = detect_state_column(frame)
    if state_column:
        state_sheets[name] = state_column
state_sheets
```

```python
sheet_name = next(iter(state_sheets))
frame = sheets[sheet_name]
state_column = state_sheets[sheet_name]
metrics = numeric_metric_columns(frame, state_column)
metrics[:10]
```

```python
metric = metrics[0]
mapped = prepare_state_metric(frame, metric, state_column)
mapped.head(10)
```

The selected measure is descriptive public data. Always verify the source worksheet and field definition before interpreting a ranking.
