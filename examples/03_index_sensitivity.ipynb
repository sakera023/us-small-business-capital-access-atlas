# Example 3 — Capital Access Index Sensitivity

Build a transparent exploratory index from a small illustrative state table and test how rankings change when a component is removed.

> This notebook uses illustrative values to demonstrate methodology. It does not publish a substantive state ranking.


```python
import pandas as pd
from capital_access_atlas import build_composite_index, leave_one_metric_out_sensitivity
```

```python
demo = pd.DataFrame({
    'State': ['VA', 'MD', 'NC', 'GA', 'TX', 'CA'],
    'Entrepreneurial Activity': [70, 68, 64, 72, 81, 85],
    'Employment Opportunity': [66, 71, 62, 69, 78, 82],
    'Access Constraint': [45, 38, 52, 48, 41, 36],
})
demo
```

```python
weights = {
    'Entrepreneurial Activity': 1.0,
    'Employment Opportunity': 1.0,
    'Access Constraint': 1.0,
}
index = build_composite_index(
    demo,
    state_column='State',
    metric_weights=weights,
    inverse_metrics={'Access Constraint'},
)
index
```

```python
sensitivity = leave_one_metric_out_sensitivity(
    demo,
    state_column='State',
    metric_weights=weights,
    inverse_metrics={'Access Constraint'},
)
sensitivity
```

A professional composite index should report component values, source years, missing-data rules, weight sensitivity, and alternative specifications—not only a single rank.
