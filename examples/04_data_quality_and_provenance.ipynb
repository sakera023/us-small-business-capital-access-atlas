# Example 4 — Data Quality and Provenance

Inspect metric completeness before using public fields in a map or composite index.


```python
import pandas as pd
from capital_access_atlas import metric_quality_report
```

```python
example = pd.DataFrame({
    'State': ['VA', 'MD', 'NC', 'GA', 'TX'],
    'Measure A': [100, 120, None, 90, 200],
    'Measure B': [0.4, 0.6, 0.5, 0.55, 0.7],
})
metric_quality_report(example, ['Measure A', 'Measure B'])
```

## Provenance checklist

Before integrating a measure, record its publisher, official URL, vintage, geography, field definition, transformation, missing-value treatment, license/public-use status, and known limitations.
